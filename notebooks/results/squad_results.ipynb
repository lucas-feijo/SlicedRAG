## Loading results (OPT-125M, 40% sparsity, pruned on SQuAD v2)

## Empty predictions

```python
percent_empty = df["prediction"].apply(lambda x: x.strip() == "").mean()
print(f"Percentage of empty predictions: {percent_empty:.2%}")
```

## Non-empty predictions

```python
not_empty = df[df["prediction"].apply(lambda x: x.strip() != "")]
```

```python
# random examples of non-empty predictions

examples = not_empty.sample(5, random_state=42)

for idx, row in examples.iterrows():
    print(f"Context: {row['context']}")
    print(f"Question: {row['question']}")
    print(f"Prediction: {row['prediction']}")
    print(f"Ground Truth: {row['target']}")
    print("-" * 80)
```
